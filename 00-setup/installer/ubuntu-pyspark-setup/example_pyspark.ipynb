# Ubuntu PySpark Lab
Runs with the **PySpark 4.1.4 (Python 3.12)** kernel.


```python
from pyspark.sql import SparkSession
spark = (SparkSession.builder.appName('Ubuntu-PySpark-Notebook')
         .master('local[2]').config('spark.sql.shuffle.partitions', '2').getOrCreate())
print('Spark version:', spark.version)
```

```python
from pathlib import Path
rows = [(1, 'Alice'), (2, 'Bob')]
df = spark.createDataFrame(rows, ['id', 'name'])
df.show()
out = Path.home() / 'spark-projects' / 'notebooks' / 'example-output'
df.write.mode('overwrite').parquet(str(out))
spark.read.parquet(str(out)).show()
```
